# Demo: Universal Cascade (V1)

Backend-only demo: один `Cascade` на automotive и counter-UAV домены.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().resolve()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

import numpy as np
from core.cascade import Cascade, DomainSpec
from ui.services.presets import get_signal

for domain_name, preset in [("automotive", "a1"), ("counter_uav", "u1")]:
    spec = DomainSpec.automotive() if domain_name == "automotive" else DomainSpec.counter_uav()
    cascade = Cascade(spec)
    sig = get_signal(preset)
    result = cascade.forward(sig, run_all_doa=True)
    print(f"{domain_name}: class={result.scene_label}, latency={result.latency_ms:.1f}ms, doa={result.doa_angles}")
```
